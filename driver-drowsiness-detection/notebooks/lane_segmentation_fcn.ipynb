# Fully Convolutional Network — lane segmentation (experiment)

VGG16-backbone FCN that segments the road area. **Not used by `src/driver_assist.py`**, which uses a classical CV pipeline (colour/gradient thresholds + sliding windows).

**Data:** the file naming (`um_`/`umm_`/`uu_`, `gt_image_2/*_road_*`) and the road label colour match the
[KITTI Road benchmark](https://www.cvlibs.net/datasets/kitti/eval_road.php) (free registration required). Put its
`training/image_2` and `training/gt_image_2` folders in `data/training/` next to this notebook. The notebook splits
whatever it finds 70 / 15 / 15 into train / validation / test.

## Load Dataset

### Source Dataset

```python
import os

# Expected layout (relative to the folder you run the notebook from):
#   data/training/image_2/     road images
#   data/training/gt_image_2/  ground-truth lane masks
#   data/testing/              test images / videos

# Option 1: Use current working directory (most common and recommended)
base_dir = os.getcwd()
data_path = os.path.join(base_dir, 'data', 'my_file.txt')
print(f"Data path using os.getcwd(): {data_path}")

# Option 2: If the notebook is in a subdirectory and you need to go "up"
# If the notebook is in a subfolder and the data folder is one level up
# parent_dir = os.path.abspath(os.path.join(os.getcwd(), os.pardir))
# data_path = os.path.join(parent_dir, 'data', 'my_file.txt')
# print(f"Data path using os.pardir: {data_path}")
data_dir = os.path.join(base_dir, "data")
train_data_dir = os.path.join(data_dir, "training", "image_2")
train_gt_dir = os.path.join(data_dir, "training", "gt_image_2")
test_data_dir = os.path.join(data_dir, "testing")

# Create directories if they don't exist
os.makedirs(train_data_dir, exist_ok=True)
os.makedirs(train_gt_dir, exist_ok=True)
os.makedirs(test_data_dir, exist_ok=True)
```

```python
def display_sample(display_list):
    plt.figure(figsize=(18, 18))
    title = ['Input Image', 'True Mask', 'Predicted Mask']
    
    for i in range(len(display_list)):
        plt.subplot(1, len(display_list), i+1)
        plt.title(title[i])
        plt.imshow(tf.keras.preprocessing.image.array_to_img(display_list[i]))
        plt.axis('off')
    
    plt.show()
    plt.close()  # Add this to prevent memory leaks
    
```

```python
# Initialize Constants
IMG_SIZE = 128
N_CHANNELS = 3
N_CLASSES = 1
SEED = 123
tf.keras.utils.set_random_seed(SEED)
```

```python
# Function to load image and return a dictionary
def parse_image(img_path: str) -> dict:
    image = tf.io.read_file(img_path)
    image = tf.io.decode_image(image, channels=3, expand_animations=False)   # jpg or png
    image = tf.image.convert_image_dtype(image, tf.uint8)

    # Three types of img paths: um, umm, uu
    # gt image paths: um_road, umm_road, uu_road
    mask_path = tf.strings.regex_replace(img_path, "image_2", "gt_image_2")
    mask_path = tf.strings.regex_replace(mask_path, "um_", "um_road_")
    mask_path = tf.strings.regex_replace(mask_path, "umm_", "umm_road_")
    mask_path = tf.strings.regex_replace(mask_path, "uu_", "uu_road_")

    mask = tf.io.read_file(mask_path)
    mask = tf.image.decode_png(mask, channels=3)

    non_road_label = np.array([255, 0, 0])
    road_label = np.array([255, 0, 255])
    other_road_label = np.array([0, 0, 0])

    # Convert to mask to binary mask
    mask = tf.experimental.numpy.all(mask == road_label, axis = 2)
    mask = tf.cast(mask, tf.uint8)
    mask = tf.expand_dims(mask, axis=-1)

    return {'image': image, 'segmentation_mask': mask}
```

```python
# Train / validation / test split, derived from the files that are actually present.
TRAIN_FRAC, VAL_FRAC = 0.70, 0.15          # the remainder is the test set

def split_files(files, seed, train_frac=TRAIN_FRAC, val_frac=VAL_FRAC):
    """Deterministic train/val/test split of a list of file paths.

    The shuffle happens ONCE, here, in plain Python. (tf.data.Dataset.list_files() reshuffles on every pass
    over the data, so take()/skip() silently moved images between train, validation and test from one epoch
    to the next.)"""
    files = sorted(files)
    random.Random(seed).shuffle(files)
    n_train = int(len(files) * train_frac)
    n_val = int(len(files) * val_frac)
    return files[:n_train], files[n_train:n_train + n_val], files[n_train + n_val:]

# Try .jpg first, fallback to .png if needed
jpg_files = tf.io.gfile.glob(os.path.join(train_data_dir, "*.jpg"))
png_files = tf.io.gfile.glob(os.path.join(train_data_dir, "*.png"))
image_files = sorted(set(jpg_files + png_files))

TRAINSET_SIZE = VALIDSET_SIZE = TESTSET_SIZE = 0
if image_files:
    train_files, val_files, test_files = split_files(image_files, SEED)
    TRAINSET_SIZE, VALIDSET_SIZE, TESTSET_SIZE = len(train_files), len(val_files), len(test_files)
    print(f"Found {len(image_files)} images -> train {TRAINSET_SIZE} / val {VALIDSET_SIZE} / test {TESTSET_SIZE}")

    def make_dataset(files):
        return tf.data.Dataset.from_tensor_slices(files).map(parse_image)

    train_dataset = make_dataset(train_files)
    val_dataset = make_dataset(val_files)
    test_dataset = make_dataset(test_files)
else:
    print(
        f"No image files found in {train_data_dir} with .jpg or .png extension.\n"
        f"Checked path: {train_data_dir}\n"
        f"Files in directory: {os.listdir(train_data_dir)}"
    )
    # Skip dataset creation to avoid error
    train_dataset = None
    val_dataset = None
    test_dataset = None
```

### Apply Transformations

```python
# Tensorflow function to rescale images to [0, 1]
@tf.function
def normalize(input_image: tf.Tensor, input_mask: tf.Tensor) -> tuple:
    input_image = tf.cast(input_image, tf.float32) / 255.0
    return input_image, input_mask

# Tensorflow function to apply preprocessing transformations
@tf.function
def load_image_train(datapoint: dict) -> tuple:
    input_image = tf.image.resize(datapoint['image'], (IMG_SIZE, IMG_SIZE))
    input_mask = tf.image.resize(datapoint['segmentation_mask'], (IMG_SIZE, IMG_SIZE), method='nearest')

    if tf.random.uniform(()) > 0.5:
        input_image = tf.image.flip_left_right(input_image)
        input_mask = tf.image.flip_left_right(input_mask)

    input_image, input_mask = normalize(input_image, input_mask)

    return input_image, input_mask

# Tensorflow function to preprocess validation images
@tf.function
def load_image_test(datapoint: dict) -> tuple:
    input_image = tf.image.resize(datapoint['image'], (IMG_SIZE, IMG_SIZE))
    input_mask = tf.image.resize(datapoint['segmentation_mask'], (IMG_SIZE, IMG_SIZE), method='nearest')

    input_image, input_mask = normalize(input_image, input_mask)

    return input_image, input_mask
```

```python
BATCH_SIZE = 32
BUFFER_SIZE = 1000

dataset = {"train": train_dataset, "val": val_dataset, "test": test_dataset}

if dataset['train'] is not None:
	# -- Train Dataset --#
	dataset['train'] = dataset['train'].map(load_image_train, num_parallel_calls=tf.data.AUTOTUNE)
	dataset['train'] = dataset['train'].shuffle(buffer_size=BUFFER_SIZE, seed=SEED)
	dataset['train'] = dataset['train'].repeat()
	dataset['train'] = dataset['train'].batch(BATCH_SIZE)
	dataset['train'] = dataset['train'].prefetch(buffer_size=tf.data.AUTOTUNE)
else:
	print("Warning: train_dataset is None. Skipping train dataset processing.")

if dataset['val'] is not None:
	#-- Validation Dataset --#
	dataset['val'] = dataset['val'].map(load_image_test)
	dataset['val'] = dataset['val'].repeat()
	dataset['val'] = dataset['val'].batch(BATCH_SIZE)
	dataset['val'] = dataset['val'].prefetch(buffer_size=tf.data.AUTOTUNE)
else:
	print("Warning: val_dataset is None. Skipping validation dataset processing.")

if dataset['test'] is not None:
	#-- Testing Dataset --#
	dataset['test'] = dataset['test'].map(load_image_test)
	dataset['test'] = dataset['test'].batch(BATCH_SIZE)
	dataset['test'] = dataset['test'].prefetch(buffer_size=tf.data.AUTOTUNE)
else:
	print("Warning: test_dataset is None. Skipping test dataset processing.")

print(dataset['train'])
print(dataset['val'])
print(dataset['test'])
```

```python
# Function to view the images from the directory
def display_sample(display_list):
    plt.figure(figsize=(18, 18))

    title = ['Input Image', 'True Mask', 'Predicted Mask']

    for i in range(len(display_list)):
        plt.subplot(1, len(display_list), i+1)
        plt.title(title[i])
        plt.imshow(tf.keras.preprocessing.image.array_to_img(display_list[i]))
        plt.axis('off')

    plt.show()

if dataset["train"] is not None:
    for image, mask in dataset["train"].take(1):
        sample_image, sample_mask = image, mask
    display_sample([sample_image[0], sample_mask[0]])
else:
    print("Warning: dataset['train'] is None. Please check your data loading step.")
```

## Define Network

```python
# Get VGG-16 network as backbone
vgg16_model = VGG16()
vgg16_model.summary()
```

```python
# Define input shape
input_shape = (IMG_SIZE, IMG_SIZE, N_CHANNELS)
```

```python
# Generate a new model using the VGG network
# Input
inputs = Input(input_shape)

# VGG network
vgg16_model = VGG16(include_top = False, weights = 'imagenet', input_tensor = inputs)

# Encoder Layers
c1 = vgg16_model.get_layer("block3_pool").output
c2 = vgg16_model.get_layer("block4_pool").output
c3 = vgg16_model.get_layer("block5_pool").output

# Decoder
u1 = UpSampling2D((2, 2), interpolation = 'bilinear')(c3)
d1 = Concatenate()([u1, c2])

u2 = UpSampling2D((2, 2), interpolation = 'bilinear')(d1)
d2 = Concatenate()([u2, c1])

# Output
u3 = UpSampling2D((8, 8), interpolation = 'bilinear')(d2)
outputs = Conv2D(N_CLASSES, 1, activation = 'sigmoid')(u3)

model = Model(inputs, outputs, name = "VGG_FCN8")
```

## Training

### Loss Function

```python
m_iou = tf.keras.metrics.BinaryIoU(target_class_ids=[1], threshold=0.5)
model.compile(optimizer=Adam(),
              loss=BinaryCrossentropy(),
              metrics=[m_iou])
```

### Check Model

```python
# Function to create a mask out of network prediction
def create_mask(pred_mask: tf.Tensor) -> tf.Tensor:
    # Round to closest
    pred_mask = tf.math.round(pred_mask)

    # [IMG_SIZE, IMG_SIZE] -> [IMG_SIZE, IMG_SIZE, 1]
    # Prediction already has shape [H, W, 1].
    return pred_mask

# Function to show predictions
def show_predictions(dataset=None, num=1):
    if dataset:
        # Predict and show image from input dataset
        for image, mask in dataset.take(num):
            pred_mask = model.predict(image)
            display_sample([image[0], mask[0], create_mask(pred_mask[0])])
    else:
        # Predict and show the sample image
        inference = model.predict(sample_image)
        display_sample([sample_image[0], sample_mask[0],
                        inference[0]])

if dataset['train'] is not None:
    for image, mask in dataset['train'].take(1):
        sample_image, sample_mask = image, mask
    show_predictions()
else:
    print("Warning: dataset['train'] is None. Please check your data loading step.")
```

### Train Model

```python
# Callbacks and Logs
class DisplayCallback(callbacks.Callback):
    def on_epoch_end(self, epoch, logs=None):
        clear_output(wait=True)
        show_predictions()
        print ('\nSample Prediction after epoch {}\n'.format(epoch+1))

logdir = os.path.join("logs", datetime.datetime.now().strftime("%Y%m%d-%H%M%S"))

training_callbacks = [
    DisplayCallback(),
    callbacks.TensorBoard(logdir, histogram_freq = -1),
    callbacks.EarlyStopping(patience = 10, verbose = 1),
    callbacks.ModelCheckpoint('best_model.h5', verbose = 1, save_best_only = True)
]

# Sizes come from the data-loading cell (derived from the files actually found).
# max(1, ...) keeps very small datasets trainable.
EPOCHS = 200
STEPS_PER_EPOCH = max(1, TRAINSET_SIZE // BATCH_SIZE)
VALIDATION_STEPS = max(1, VALIDSET_SIZE // BATCH_SIZE)
```

```python
if dataset['train'] is not None and dataset['val'] is not None:
    model_history = model.fit(dataset['train'], epochs=EPOCHS,
                              steps_per_epoch=STEPS_PER_EPOCH,
                              validation_data=dataset["val"],
                              validation_steps=VALIDATION_STEPS,
                              callbacks=training_callbacks)
else:
    print("Error: Training or validation dataset is None. Please check your data loading step.")
```

## Testing (Test Dataset)

```python
# Function to calculate mask over image
def weighted_img(img, initial_img, α=1., β=0.5, γ=0.):
    return cv2.addWeighted(initial_img, α, img, β, γ)

# Function to process an individual image and it's mask
def process_image_mask(image, mask):
    # Round to closest
    mask = tf.math.round(mask)

    # Convert to mask image
    zero_image = np.zeros_like(mask)
    mask = np.dstack((mask, zero_image, zero_image))
    mask = np.asarray(mask, np.float32)

    # Convert to image image
    image = np.asarray(image, np.float32)

    # Get the final image
    final_image = weighted_img(mask, image)

    return final_image
```

```python
# Function to save predictions
def save_predictions(dataset):
    # Predict and save image the from input dataset
    index = 0
    for batch_image, batch_mask in dataset:
        for image, mask in zip(batch_image, batch_mask):
            print(f"Processing image : {index}")
            pred_mask = model.predict(tf.expand_dims(image, axis = 0))
            save_sample([image, process_image_mask(image, pred_mask[0])], index)
            index += 1

# Function to save the images as a plot
def save_sample(display_list, index):
    plt.figure(figsize=(18, 18))

    title = ['Input Image', 'Predicted Mask']

    for i in range(len(display_list)):
        plt.subplot(1, len(display_list), i+1)
        plt.title(title[i])
        plt.imshow(tf.keras.preprocessing.image.array_to_img(display_list[i]))
        plt.axis('off')

    plt.savefig(f"outputs/{index}.png")
    plt.show()
```

## Testing (Videos)

```python
# Function to view video
class HTML:
    def __init__(self, html):
        self.html = html

    def __str__(self):
        raise NotImplementedError

    def __repr__(self):
        raise NotImplementedError

def play(filename):
    html = ''
    video = open(filename,'rb').read()
    src = 'data:video/mp4;base64,' + b64encode(video).decode()
    html += '<video width=1000 controls autoplay loop><source src="%s" type="video/mp4"></video>' % src
    return HTML(html)
```

```python
# Function to process an individual image
def process_image(image):
    # Preprocess image
    original_size = (image.shape[1], image.shape[0])
    image = cv2.resize(image, (IMG_SIZE, IMG_SIZE))
    rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB).astype(np.float32) / 255.0
    # Match training RGB channel order and [0, 1] input range.
    pred_mask = model.predict(np.expand_dims(rgb, axis=0))
    mask = np.round(pred_mask[0])

    # Convert to mask image
    zero_image = np.zeros_like(mask)
    mask = np.dstack((mask, zero_image, zero_image)) * 255
    mask = np.asarray(mask, np.uint8)

    # Get the final image
    final_image = weighted_img(mask, image)
    final_image = cv2.resize(final_image, original_size)

    return final_image
```
